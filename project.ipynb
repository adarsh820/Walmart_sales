```python
#importing library
import pandas as pd

#reading the dataset
import pymysql
from sqlalchemy import create_engine
```

```python
df =pd.read_csv('Walmart.csv')
```

```python
df.fillna(0, inplace= True)
```

```python
#replacing $ sign in unit_price with noting and converting the column into float type

df['unit_price']= df['unit_price'].str.replace('$','').astype(float)
```

```python
#creating a new column 'total_price' by multiplying 'unit_price' and 'quantity'
df['total_price']= df['unit_price']* df['quantity']
```
